```python
# Sort strings by their length instead of alphabetically
words = ["apple", "pie", "banana", "kiwi"]
sorted_words = sorted(words, key=len)
print(sorted_words) # ['pie', 'kiwi', 'apple', 'banana']
```

```python
# Sort case-insensitively using str.lower method reference
names = ["Alice", "bob", "Charlie", "david"] # case example
names_sorted = sorted(["Alice", "bob", "Charlie", "david"], key=str.lower)
print(names_sorted)
```

```python
# List of dictionaries representing items with prices
items = [
    {"name": "laptop", "price": 1200},
    {"name": "mouse", "price": 25},
    {"name": "keyboard", "price": 75}
]

# Sort items by price using a lambda expression
cheapest_first = sorted(items, key=lambda x: x["price"])
print(cheapest_first)
```

```python
# multi-key sorting
students = [
    {"name": "Alice", "grade": 85, "age": 20},
    {"name": "Bob", "grade": 90, "age": 19},
    {"name": "Charlie", "grade": 85, "age": 21}
]

# Sort primarily by grade (ascending), 
# then by age (ascending) if grades match
sorted_students = sorted(
    students, 
    key=lambda s: (s["grade"], s["age"])
)
print(sorted_students)
```

```python
# using operator to select attributes
from operator import itemgetter, attrgetter

# Using itemgetter instead of lambda for dictionary keys
sorted_items = sorted(items, key=itemgetter("price"))

# Reversing order functionally
descending_items = sorted(items, key=itemgetter("price"), reverse=True)
print(descending_items)
```
